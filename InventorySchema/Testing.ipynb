```python
from Connection.connection_manager import ConnectionManager

# Create connection manager
manager = ConnectionManager()

# Start Spark using platform_config.yaml
spark_result = manager.connect_spark()

# Check connection
print(spark_result.message)

if spark_result.is_connected:

    # Get actual SparkSession
    spark = spark_result.connection

    print("\nSpark configuration:")
    print(
        "App Name       :",
        spark.conf.get("spark.app.name")
    )

    print(
        "Driver Memory  :",
        spark.conf.get("spark.driver.memory")
    )

    print(
        "Executor Memory:",
        spark.conf.get("spark.executor.memory")
    )

    print(
        "Executor Cores :",
        spark.conf.get("spark.executor.cores")
    )

    print(
        "Shuffle Partitions:",
        spark.conf.get(
            "spark.sql.shuffle.partitions"
        )
    )

    print(
        "S3A Endpoint   :",
        spark.conf.get(
            "spark.hadoop.fs.s3a.endpoint"
        )
    )

    print("\nSpark test:")
    spark.range(5).show()

else:

    print("Spark connection failed.")
    print(spark_result.error)
```
