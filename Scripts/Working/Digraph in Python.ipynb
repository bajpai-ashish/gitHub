```python
from graphviz import Digraph

dot = Digraph(comment="PySpark Pipeline")

# Define nodes
dot.node("E", "Extract")
dot.node("T", "Transform")
dot.node("L", "Load")

# Define edges
dot.edges(["ET", "TL"])

# Save diagram
dot.render("pipeline_diagram.gv", view=True)
```

```python
from graphviz import Digraph
dot = Digraph()
dot.node("A", "Extract")
dot.node("B", "Transform")
dot.edge("A", "B")
dot.render("test_pipeline.gv", view=True)
```

```python
from graphviz import Digraph

dot = Digraph(comment="Pipeline DAG")

dot.node("E", "Extract", shape="box", style="filled", fillcolor="lightblue")
dot.node("T", "Transform", shape="ellipse", style="filled", fillcolor="lightgreen")
dot.node("L", "Load", shape="diamond", style="filled", fillcolor="orange")

dot.edge("E", "T", color="blue")
dot.edge("T", "L", color="red")

dot.render("pipeline_colored.gv", view=True)
```

```python
status_colors = {
    "Completed": "green",
    "Running": "yellow",
    "Failed": "red"
}

dot.node("Extract", style="filled", fillcolor=status_colors["Completed"])
dot.node("Transform", style="filled", fillcolor=status_colors["Running"])
dot.node("Load", style="filled", fillcolor=status_colors["Failed"])
```

```python
import networkx as nx
import matplotlib.pyplot as plt

# Create graph
G = nx.DiGraph()

# Add nodes with metadata
G.add_node("Extract", status="Completed")
G.add_node("Transform", status="Completed")
G.add_node("Load", status="Pending")

# Add edges
G.add_edges_from([("Extract", "Transform"), ("Transform", "Load")])

# Draw graph
pos = nx.spring_layout(G)
nx.draw(G, pos, with_labels=True, node_color="lightblue", node_size=2000, font_size=12)

# Annotate with status
labels = nx.get_node_attributes(G, "status")
nx.draw_networkx_labels(G, pos, labels=labels)

plt.savefig("pipeline_networkx.png")
```
