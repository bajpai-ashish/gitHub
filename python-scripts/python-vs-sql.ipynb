### <center> SQL </center>

```python
# we need to update each column individually

conn = pgs.connect(database = "Education_data", 
                        user = "postgres", 
                        host= 'localhost',
                        password = "****",
                        port = 5432)

cur = conn.cursor()
sql4 = ''' UPDATE gross_enrollment 
SET ge_hrsecondary_total = CASE  
  WHEN ge_hrsecondary_total = 'NR' THEN null 
  WHEN ge_hrsecondary_total = '@' THEN null   
  ELSE ge_hrsecondary_total  
END 
WHERE ge_hrsecondary_total = 'NR' or ge_hrsecondary_total = '@';'''

cur.execute(sql4)

sql5 = ''' UPDATE gross_enrollment 
SET ge_hrsecondary_girls = CASE  
  WHEN ge_hrsecondary_girls = 'NR' THEN null 
  WHEN ge_hrsecondary_girls = '@' THEN null   
  ELSE ge_hrsecondary_girls  
END 
WHERE ge_hrsecondary_girls = 'NR' or ge_hrsecondary_girls = '@';'''

cur.execute(sql5)

sql6 = ''' UPDATE gross_enrollment 
SET ge_hrsecondary_boys = CASE  
  WHEN ge_hrsecondary_boys = 'NR' THEN null 
  WHEN ge_hrsecondary_boys = '@' THEN null   
  ELSE ge_hrsecondary_boys  
END 
WHERE ge_hrsecondary_boys = 'NR' or ge_hrsecondary_boys = '@';'''

cur.execute(sql6)

conn.commit()
conn.close()
```

## 8. Add a new column based on values of an existing column

### <center> Python </center>

```python
#defining a function that we will use to map the values in new column

def zone(value):
    if value in ('Haryana','Himachal Pradesh','Rajasthan','Punjab','Delhi','Jammu And Kashmir','Chandigarh','Uttarakhand'):
        return "North"
    elif value in ('Arunachal Pradesh','Assam','Manipur','Meghalaya','Mizoram','Nagaland','Sikkim','Tripura'):
        return "North_east"
    elif value in ('Uttar Pradesh','Madhya Pradesh','Chhattisgarh'):
        return "Central"
    elif value in ('Gujarat','Maharashtra','Daman & Diu','Dadra & Nagar Haveli','Goa',):
        return "West"
    elif value in ('Bihar','Odisha','West Bengal','Jharkhand'):
        return "East"
    elif value in ('Andhra Pradesh','Tamil Nadu','Kerala','Telangana','Karnataka','Puducherry'):
        return "South"
    elif value in ('Lakshadweep','Andaman & Nicobar Islands'):
        return "Islands"
    else :
        return "N/A"
```

```python
# mapping the new column to all the dataframes
df_list_2 = [merged_df, merged_df_2, gross_enrollment, drop_out]
for i in df_list_2:
    i['Zone'] = i['State_UT'].map(zone)


merged_df.info()
```
