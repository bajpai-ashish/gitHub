# One common Spark connection

mermaid
```
flowchart TD
    spk[Spark 4.0.1] -->|Read| csv(Raw Bucket)
    csv -->|Load| Hudi[Hudi]
    csv -->|Load| Delta[Delta]
    csv -->|Load| Iceberg[Iceberg]
    Hudi --> HudiE[Hudi Extensions]
    Delta --> DeltaE[Delta Extensions]
    Iceberg --> IcebergE[Iceberg Extensions]
    DeltaE --> DeltaC[Delta Catalog]
    IcebergE --> IcebergC[Iceberg Catalog]
    HudiE --> Hudi_Buckets[Bronze Buckets]
    DeltaC --> Delta_Buckets[Silver Buckets]
    IcebergC --> Iceberg_Buckets[Gold Buckets]
```



## All 3 connections

````python
from pyspark.sql import SparkSession

# ============================================================
# Paths
# ============================================================

RAW_PATH    = "s3a://rawload"
BRONZE_PATH = "s3a://bronzeload"
SILVER_PATH = "s3a://silverload"
GOLD_PATH   = "s3a://goldload"

RAW_FILE = (
    f"{RAW_PATH}/agesexbyethnicgroup/Data8277.csv"
)

HUDI_PATH = f"{BRONZE_PATH}/agesexbyethnicgroup"
DELTA_PATH = f"{SILVER_PATH}/agesexbyethnicgroup"

# ============================================================
# Spark
# ============================================================

spark = (
    SparkSession.builder
    .appName("Three Table Format Comparison")

    # --------------------------------------------------------
    # Local Hudi bundle
    # --------------------------------------------------------

    .config(
        "spark.jars",
        r"C:\data\datatool\spark\jars"
        r"\hudi-spark4.0-bundle_2.13-1.2.0.jar"
    )

    # --------------------------------------------------------
    # Delta + Iceberg + Hadoop AWS
    # --------------------------------------------------------

    .config(
        "spark.jars.packages",
        "io.delta:delta-spark_2.13:4.0.0,"
        "org.apache.iceberg:iceberg-spark-runtime-4.0_2.13:1.10.0,"
        "org.apache.hadoop:hadoop-aws:3.4.1"
    )

    # --------------------------------------------------------
    # Spark SQL extensions
    # --------------------------------------------------------
    .config(
        "spark.sql.extensions",
        "org.apache.spark.sql.hudi.HoodieSparkSessionExtension,"
        "io.delta.sql.DeltaSparkSessionExtension,"
        "org.apache.iceberg.spark.extensions.IcebergSparkSessionExtensions"
        )

    # --------------------------------------------------------
    # Delta catalog
    # --------------------------------------------------------

    .config(
        "spark.sql.catalog.spark_catalog",
        "org.apache.spark.sql.delta.catalog.DeltaCatalog"
    )

    # --------------------------------------------------------
    # Hudi
    # --------------------------------------------------------

    .config(
        "spark.serializer",
        "org.apache.spark.serializer.KryoSerializer"
    )

    .config(
        "spark.kryo.registrator",
        "org.apache.spark.HoodieSparkKryoRegistrar"
    )

    # --------------------------------------------------------
    # Iceberg catalog
    # --------------------------------------------------------

    .config(
        "spark.sql.catalog.ice",
        "org.apache.iceberg.spark.SparkCatalog"
    )

    .config(
        "spark.sql.catalog.ice.catalog-impl",
        "org.apache.iceberg.hadoop.HadoopCatalog"
    )

    .config(
        "spark.sql.catalog.ice.warehouse",
        GOLD_PATH
    )

    # --------------------------------------------------------
    # MinIO / S3A
    # --------------------------------------------------------

    .config(
        "spark.hadoop.fs.s3a.endpoint",
        "http://127.0.0.1:9000"
    )

    .config(
        "spark.hadoop.fs.s3a.access.key",
        "minioadmin"
    )

    .config(
        "spark.hadoop.fs.s3a.secret.key",
        "minioadmin"
    )

    .config(
        "spark.hadoop.fs.s3a.path.style.access",
        "true"
    )

    .config(
        "spark.hadoop.fs.s3a.impl",
        "org.apache.hadoop.fs.s3a.S3AFileSystem"
    )

    .config(
        "spark.hadoop.fs.s3a.aws.credentials.provider",
        "org.apache.hadoop.fs.s3a.SimpleAWSCredentialsProvider"
    )

    .getOrCreate()
)

print("Spark :", spark.version)

print(
    "Hadoop:",
    spark.sparkContext._jvm
        .org.apache.hadoop.util.VersionInfo.getVersion()
)
````

# Read CSV ONCE

````python
from pyspark.sql import functions as F

raw_df = (
    spark.read
    .option("header", "true")
    .option("inferSchema", "false")
    .csv(RAW_FILE)
)

print("RAW rows:", raw_df.count())

raw_df.printSchema()

raw_df.show(5, truncate=False)
````

## create a technical record ID for Hudi.

````python
df = raw_df.withColumn(
    "_record_id",
    F.expr("uuid()")
)

# df = df.persist()

print("Cached rows:", df.count())
````

# The SAME DataFrame to all three

## Hudi only

````python
import time

start = time.perf_counter()

(
    df.write
    .format("org.apache.hudi")
    .option(
        "hoodie.table.name",
        "agesexbyethnicgroup"
    )
    .option(
        "hoodie.datasource.write.operation",
        "bulk_insert"
    )
    .option(
        "hoodie.datasource.write.recordkey.field",
        "_record_id"
    )
    .option(
        "hoodie.datasource.write.precombine.field",
        "_record_id"
    )
    .mode("overwrite")
    .save(HUDI_PATH)
)

hudi_time = time.perf_counter() - start

print(f"Hudi bulk_insert time: {hudi_time:.2f} seconds")
````

### Read from Hudi

````python
hudi_df = (
    spark.read
    .format("hudi")
    .load(HUDI_PATH)
)

print("Hudi rows:", hudi_df.count())
hudi_df.show(5, truncate=False)
````

## Delta only

````python
import time

start = time.perf_counter()

(
    df.write
    .format("delta")
    .mode("overwrite")
    .save(DELTA_PATH)
)

delta_time = time.perf_counter() - start

print(f"Delta write time: {delta_time:.2f} seconds")
````

### Read from Delta

````python
delta_df = (
    spark.read
    .format("delta")
    .load(DELTA_PATH)
)

print("Delta rows:", delta_df.count())
delta_df.show(5, truncate=False)
````

## Iceberg only

````python
import time

start = time.perf_counter()

(
    df.writeTo("ice.agesexbyethnicgroup")
    .using("iceberg")
    .createOrReplace()
)

iceberg_time = time.perf_counter() - start

print(f"Iceberg write time: {iceberg_time:.2f} seconds")
````

### Read from iceberg

````python
iceberg_df = spark.table("ice.agesexbyethnicgroup")

print("Iceberg rows:", iceberg_df.count())
iceberg_df.show(5)
````
