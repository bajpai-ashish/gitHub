```python
from pyspark.sql import functions as F

raw_file = (
    f"{RAW_PATH}/"
    "machine-readable-business-employment-data-mar-2026-quarter.csv"
)

raw_df = (
    spark.read
    .option("header", "true")
    .option("inferSchema", "false")
    .csv(raw_file)
)

print("RAW rows:", raw_df.count())

raw_df.show(5, truncate=False)
```

```python
bronze_df = raw_df.withColumn(
    "_record_id",
    F.expr("uuid()")
)
```

```python
hudi_path = f"{BRONZE_PATH}/business_employment"

(
    bronze_df.write
    .format("org.apache.hudi")

    .option(
        "hoodie.table.name",
        "business_employment"
    )

    .option(
        "hoodie.datasource.write.table.type",
        "COPY_ON_WRITE"
    )

    .option(
        "hoodie.datasource.write.recordkey.field",
        "_record_id"
    )

    .option(
        "hoodie.datasource.write.precombine.field",
        "_record_id"
    )

    .mode("overwrite")

    .save(hudi_path)
)

print("Hudi Bronze created.")
```

```python
hudi_df = (
    spark.read
    .format("hudi")
    .load(hudi_path)
)

print("Hudi rows:", hudi_df.count())

hudi_df.show(5, truncate=False)
```

```python
print("RAW :", raw_df.count())
print("HUDI:", hudi_df.count())
```

# Delta environment

```python
from pyspark.sql import SparkSession

RAW_PATH    = "s3a://rawload"
BRONZE_PATH = "s3a://bronzeload"
SILVER_PATH = "s3a://silverload"
GOLD_PATH   = "s3a://goldload"

spark = (
    SparkSession.builder
    .appName("Delta Silver Test")

    # ========================================================
    # Delta Lake
    # ========================================================

    .config(
        "spark.jars.packages",
        "io.delta:delta-spark_2.13:4.0.0,"
        "org.apache.hadoop:hadoop-aws:3.4.1"
    )

    .config(
        "spark.sql.extensions",
        "io.delta.sql.DeltaSparkSessionExtension"
    )

    .config(
        "spark.sql.catalog.spark_catalog",
        "org.apache.spark.sql.delta.catalog.DeltaCatalog"
    )

    # ========================================================
    # MinIO / S3A
    # ========================================================

    .config(
        "spark.hadoop.fs.s3a.endpoint",
        "http://127.0.0.1:9000"
    )

    .config(
        "spark.hadoop.fs.s3a.access.key",
        "minioadmin"
    )

    .config(
        "spark.hadoop.fs.s3a.secret.key",
        "minioadmin"
    )

    .config(
        "spark.hadoop.fs.s3a.path.style.access",
        "true"
    )

    .config(
        "spark.hadoop.fs.s3a.impl",
        "org.apache.hadoop.fs.s3a.S3AFileSystem"
    )

    .config(
        "spark.hadoop.fs.s3a.aws.credentials.provider",
        "org.apache.hadoop.fs.s3a.SimpleAWSCredentialsProvider"
    )

    .getOrCreate()
)

print("Spark :", spark.version)

print(
    "Hadoop:",
    spark.sparkContext._jvm
        .org.apache.hadoop.util.VersionInfo.getVersion()
)
```

## Verify Delta

## Hudi → Delta

```python
hudi_path = "s3a://bronzeload/business_employment"

bronze_df = (
    spark.read
    .format("hudi")
    .load(hudi_path)
)

print("Bronze rows:", bronze_df.count())

bronze_df.show(5, truncate=False)
```

```python
delta_path = "s3a://silverload/business_employment"

(
    bronze_df.write
    .format("delta")
    .mode("overwrite")
    .save(delta_path)
)

print("Delta Silver table created.")
```

## Read Delta back

```python
silver_df = (
    spark.read
    .format("delta")
    .load(delta_path)
)

print("Silver rows:", silver_df.count())

silver_df.show(5, truncate=False)
```

```python
print("Bronze Hudi :", bronze_df.count())
print("Silver Delta:", silver_df.count())
```

# Spark + Iceberg + MinIO

```python
from pyspark.sql import SparkSession

RAW_PATH    = "s3a://rawload"
BRONZE_PATH = "s3a://bronzeload"
SILVER_PATH = "s3a://silverload"
GOLD_PATH   = "s3a://goldload"

spark = (
    SparkSession.builder
    .appName("Iceberg Gold Test")

    # ========================================================
    # Iceberg
    # ========================================================

    .config(
        "spark.jars.packages",
        "org.apache.iceberg:iceberg-spark-runtime-4.0_2.13:1.10.0,"
        "io.delta:delta-spark_2.13:4.0.0,"
        "org.apache.hadoop:hadoop-aws:3.4.1"
        )

    .config(
        "spark.sql.extensions",
        "org.apache.iceberg.spark.extensions.IcebergSparkSessionExtensions,"
        "io.delta.sql.DeltaSparkSessionExtension"
        )

    # ========================================================
    # Iceberg Hadoop Catalog
    # ========================================================

    .config(
        "spark.sql.catalog.ice",
        "org.apache.iceberg.spark.SparkCatalog"
    )

    .config(
        "spark.sql.catalog.ice.catalog-impl",
        "org.apache.iceberg.hadoop.HadoopCatalog"
    )

    .config(
        "spark.sql.catalog.ice.warehouse",
        GOLD_PATH
    )
    # ========================================================
    # Iceberg Delta Catalog
    # ========================================================

    .config(
        "spark.sql.catalog.spark_catalog",
        "org.apache.spark.sql.delta.catalog.DeltaCatalog"
        )

    # ========================================================
    # MinIO / S3A
    # ========================================================

    .config(
        "spark.hadoop.fs.s3a.endpoint",
        "http://127.0.0.1:9000"
    )

    .config(
        "spark.hadoop.fs.s3a.access.key",
        "minioadmin"
    )

    .config(
        "spark.hadoop.fs.s3a.secret.key",
        "minioadmin"
    )

    .config(
        "spark.hadoop.fs.s3a.path.style.access",
        "true"
    )

    .config(
        "spark.hadoop.fs.s3a.impl",
        "org.apache.hadoop.fs.s3a.S3AFileSystem"
    )

    .config(
        "spark.hadoop.fs.s3a.aws.credentials.provider",
        "org.apache.hadoop.fs.s3a.SimpleAWSCredentialsProvider"
    )

    .getOrCreate()
)

print("Spark :", spark.version)

print(
    "Hadoop:",
    spark.sparkContext._jvm
        .org.apache.hadoop.util.VersionInfo.getVersion()
)
```
